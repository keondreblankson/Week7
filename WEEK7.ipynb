Week 7 Assignment 2: Movie Ratings

For this assignment, I used five imaginary people to rate six movies on a scale of 1 to 5. These are fictional ratings. Missing ratings mean the person has not seen that movie.

```python
import pandas as pd
from IPython.display import display

data = {
    "User": ["Alex", "Jordan", "Taylor", "Morgan", "Casey"],
    "Sinners": [5, 4, 5, 3, 4],
    "Superman": [4, 3, 4, 2, 5],
    "A Minecraft Movie": [2, 2, 3, 1, None],
    "Lilo & Stitch": [4, 3, None, 2, 4],
    "F1": [5, None, 4, 3, 5],
    "The Fantastic Four: First Steps": [3, 2, 4, None, 3]
}

source_data = pd.DataFrame(data)

source_data.to_csv("movie_ratings.csv", index=False)
```

1. Load the Ratings

I saved the ratings in a CSV file and loaded that file into a pandas dataframe. Each row represents a person, and each column represents a movie. NaN means the person has not seen the movie.

```python
ratings = pd.read_csv("movie_ratings.csv", index_col="User")

display(ratings)
```

2. Average Ratings

I calculated the average rating for each person and each movie. Missing ratings are excluded from the averages instead of being counted as zero.

```python
user_averages = ratings.mean(axis=1)

movie_averages = ratings.mean(axis=0)

print("Average rating for each user:")
display(user_averages.to_frame("Average Rating").round(3))

print("Average rating for each movie:")
display(movie_averages.to_frame("Average Rating").round(3))
```

3. Normalized Ratings

I used min-max normalization for each person. I subtracted that person's lowest rating from each rating, then divided by the difference between their highest and lowest ratings. This puts their ratings on a scale from 0 to 1. Missing ratings stay missing.

```python
user_min = ratings.min(axis=1)
user_max = ratings.max(axis=1)

user_range = user_max - user_min

# Avoid dividing by zero if a person gives identical ratings.
user_range = user_range.replace(0, float("nan"))

normalized = ratings.sub(user_min, axis=0).div(user_range, axis=0)

print("Normalized ratings:")
display(normalized.round(3))

print("Average normalized rating for each user:")
display(
    normalized.mean(axis=1)
    .to_frame("Average Normalized Rating")
    .round(3)
)

print("Average normalized rating for each movie:")
display(
    normalized.mean(axis=0)
    .to_frame("Average Normalized Rating")
    .round(3)
)
```

4. Conclusion

From what I understand, normalized ratings make it easier to compare people who rate movies differently. Some people give higher ratings most of the time, while others give lower ratings. Normalization helps show which movies each person liked more compared to their other ratings.

A disadvantage is that normalized ratings lose the meaning of the original numbers. A normalized rating of 1 means it was that person's highest rating, but it does not necessarily mean they gave the movie a 5. It can also make small differences look bigger when someone uses a narrow rating range. Adding a new highest or lowest rating would change that person's normalized scores.

Overall, I would use the original ratings to show the actual scores and normalized ratings to compare people's relative preferences. Normalization does not solve the issue of missing ratings or movies having different numbers of ratings.

5. Standardized Ratings — Extra Credit

I standardized each person's ratings by subtracting their average rating and dividing by their standard deviation. I used the population standard deviation with ddof=0. A positive value means the rating was above that person's average, and a negative value means it was below their average. Each person's standardized average should be approximately zero.

```python
user_mean = ratings.mean(axis=1)
user_std = ratings.std(axis=1, ddof=0)

# Avoid dividing by zero if a person gives identical ratings.
user_std = user_std.replace(0, float("nan"))

standardized = ratings.sub(user_mean, axis=0).div(user_std, axis=0)

print("Standardized ratings:")
display(standardized.round(3))

print("Average standardized rating for each user:")
display(
    standardized.mean(axis=1)
    .to_frame("Average Standardized Rating")
    .round(3)
)

print("Average standardized rating for each movie:")
display(
    standardized.mean(axis=0)
    .to_frame("Average Standardized Rating")
    .round(3)
)
```
